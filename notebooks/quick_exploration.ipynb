# Quick exploration — interval operations
This notebook loads a small sample of the canonical fact and shows example analyses (top SLA offenders, SLA heatmap). Run after creating and activating a virtual environment.

```python
import os
import pandas as pd
import plotly.express as px
```

```python
sample_path = 'data/sample/interval_operations_sample.csv'
full_path = 'data/processed/interval_operations_analysis.csv'
if os.path.exists(sample_path):
    df = pd.read_csv(sample_path)
elif os.path.exists(full_path):
    # load a slice to keep memory usage reasonable for quick exploration
    df = pd.read_csv(full_path, nrows=5000)
else:
    raise FileNotFoundError('No sample or processed fact file found. Run the pipeline or add a sample to data/sample/')

df.head()
```

```python
# Top 20 store-interval pairs by mean SLA breach rate
agg = df.groupby(['Store_ID','Interval_ID'], as_index=False).agg({'SLA_Breach_Rate':'mean','Orders':'sum'})
agg = agg.sort_values('SLA_Breach_Rate', ascending=False)
agg.head(20)
```

```python
# SLA breach heatmap (store x interval) — may be large; subsample or aggregate if needed
pivot = df.pivot_table(index='Store_ID', columns='Interval_ID', values='SLA_Breach_Rate', aggfunc='mean').fillna(0)
fig = px.imshow(pivot, labels=dict(x='Interval_ID', y='Store_ID', color='SLA_Breach_Rate'), aspect='auto', color_continuous_scale='RdYlGn_r')
fig.update_layout(height=800)
fig.show()
```
